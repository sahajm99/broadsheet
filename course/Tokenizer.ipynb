```python
import re
import os
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

word_dict = {}
doc_dict = {}
token_id = 1
doc_id = 1

with open('stopwordlist.txt', 'r') as f:
    stop_words = {line.strip() for line in f}

def tokenize(text):
    tokens = re.findall(r'\b[a-zA-Z]+\b', text.lower())
    tokens = [token for token in tokens if token not in stop_words]
    return tokens

def stem(tokens):
    return [stemmer.stem(token) for token in tokens]

def parser(directory):
    global token_id, doc_id
    files = os.listdir(directory)
    total_files = len(files)
    for idx, filename in enumerate(files):
        filepath = os.path.join(directory, filename)
        with open(filepath, 'r') as file:
            data = file.read()
            documents = re.findall(r'<DOC>(.*?)</DOC>', data, re.DOTALL)

            for doc in documents:
                doc_no = re.search(r'<DOCNO>(.*?)</DOCNO>', doc).group(1).strip()
                
                if doc_no not in doc_dict:
                    doc_dict[doc_no] = doc_id
                    doc_id += 1
                
                text = re.search(r'<TEXT>(.*?)</TEXT>', doc, re.DOTALL).group(1).strip()
                tokens = tokenize(text)
                stemmed_tokens = stem(tokens)
                
                for token in stemmed_tokens:
                    if token not in word_dict:
                        word_dict[token] = token_id
                        token_id += 1
    
    print("Documents parsed")

def output_generator():
    with open("parser_output.txt", "w") as output_file:
        output_file.write('Token -> TokenID\n')
        for word, tid in word_dict.items():
            output_file.write(f"{word}\t{tid}\n")
        output_file.write('\nDocument -> DocumentID\n')
        for docname, did in doc_dict.items():
            output_file.write(f"{docname}\t{did}\n")

if __name__ == "__main__":
    parser('ft911')  
    output_generator()
    print("\nparser_output.txt file created containing list of tokens, documents with unique token and document ids respectively.")
```
